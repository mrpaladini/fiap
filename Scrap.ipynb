```python
# Importando as Bibliotecas que serão utilizadas

import requests
from bs4 import BeautifulSoup
import pandas as pd
import csv
import time
from sqlalchemy import create_engine
pd.set_option('display.max_colwidth', None)
```

```python
# URL base do projeto
URL = "http://books.toscrape.com/"

# Lendo a página
response = requests.get(URL)
if response.status_code == 200:
    print("Deu tudo certo segue o jogo....")
else:
    print("Deu ruim, veja oq ue precisa ser consertado...")
```

```python
# Criando o parse da página
soup = BeautifulSoup(response.text,"html.parser")
```

```python
# Extraindo as informações recebidas da página 1
books = soup.find_all('h3')
# books = soup.find_all('li')
```

```python
#Capturando o tempo de execução das tarefas
start_time = time.time()
books_extracted = 0
```

```python
# Capturando informações dos livros da primeira página

contador = 0

books_data = []

for book in books:
    book_url = book.find('a')['href']
    print(book_url)
    book_response = requests.get(URL + book_url)
    print(book_response)
    book_soup = BeautifulSoup(book_response.content, "html.parser")
    book_id = contador
    title = book_soup.find('h1').text
    category = book_soup.find('ul', class_="breadcrumb").find_all('a')[2].text.strip()
    rating = book_soup.find('p', class_='star-rating')['class'][1]
    price = book_soup.find('p', class_='price_color').text.strip().replace('£','')
    availability = book_soup.find('p', class_='availability').text.strip()
    # image = book_soup.find('div', class_='item active').find_all('img src')[2].text.strip()
    # images = book_soup.find('div', class_='item active').find_all('img').replace('../../', '')
    img_src = book_soup.find('img')['src'].replace('../../', '')
    image_url = URL + img_src

    # Imprimindo resultados
    print('*****************************************')
    print(f'Book ID:  {book_id}')
    print(f'Titulo:  {title}')
    print(f'Categoria:  {category}')
    print(f'Pontuação:  {rating}')
    print(f'Preço:   {price}')
    print(f'Disponibilidade:  {availability}')
    print(image_url)
    
    print('*****************************************')
    contador += 1
    books_extracted += 1

    books_data.append([book_id, title, category, rating, price, availability, image_url])
```

```python
# Lendo todas as páginas

for page_num in range(2, 51): # Colocar final 51 para final
    URL = f'https://books.toscrape.com/catalogue/page-{page_num}.html'
    response = requests.get(URL)
    soup = BeautifulSoup(response.content, 'html.parser')
    print(URL)
    
    # *****************

    books = soup.find_all('h3')
    books_extracted += 0
    URL_IMG = URL
    
    for book in books:
        book_url = book.find('a')['href']
        book_response = requests.get('https://books.toscrape.com/catalogue/' + book_url)
        book_soup = BeautifulSoup(book_response.content, "html.parser")
        book_id = contador
        title = book_soup.find('h1').text
        category = book_soup.find('ul', class_="breadcrumb").find_all('a')[2].text.strip()
        rating = book_soup.find('p', class_='star-rating')['class'][1]
        price = book_soup.find('p', class_='price_color').text.strip().replace('£','')
        availability = book_soup.find('p', class_='availability').text.strip()
        img_src = book_soup.find('img')['src'].replace('../../', '')
        image_url = 'https://books.toscrape.com/' + img_src
        books_data.append([book_id, title, category, rating, price, availability, image_url]) 
        # print(image_url)     
        books_extracted += 1
        contador += 1
        
end_time = time.time()
total_time = (end_time - start_time)/60.0

print("Total de Livros = "  + str(contador))
print( "Tempo total = " + str(total_time))
    
        
```

```python
# Exportar os dados
```

```python
df = pd.DataFrame(books_data, columns=["Book ID", "Title", "Category", "Rating", "Price", "Availability", "URL"])
# Test
print(df.head(10))

df.to_csv("books_final.csv", index=False)
```

```python
# Exportar para MySQL

# Conexão
usuario = 'root'
senha = "Password"
host = '10.0.2.15'
banco_de_dados = 'books' 
tabela = 'books_final'
```

```python
# Criando a string de conexão (SQLAlchemy Engine)
# Formato: mysql+mysqlconnector://usuario:senha@host/nome_do_banco
engine = create_engine(f'mysql+mysqlconnector://{usuario}:{senha}@{host}/{banco_de_dados}')
```

```python
try:
    # Exportando o DataFrame para o MySQL
    # if_exists='replace' substitui a tabela se ela já existir
    # index=False evita que o índice do Pandas vire uma coluna no banco [3]
    df.to_sql(tabela, con=engine, if_exists='replace', index=False)
    print(f"Dados exportados com sucesso para a tabela '{tabela}' no MySQL!")
except Exception as e:
    print(f"Erro ao exportar para o banco de dados: {e}")
```
